# BatteryTwin: dataset inspection and EDA
Real NASA discharge cycles; run download_data.py and prepare_data.py first.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
sys.path.insert(0,str(ROOT))
import pandas as pd
import matplotlib.pyplot as plt
```

```python
cycles = pd.read_csv(ROOT/'data/processed/cycles.csv')
cycles.groupby('battery_id').agg(cycles=('cycle','size'),first_capacity=('capacity_ah','first'),last_capacity=('capacity_ah','last'))
```

```python
for battery, group in cycles.groupby('battery_id'):
    plt.plot(group.cycle,group.capacity_ah,label=battery)
plt.xlabel('Discharge cycle'); plt.ylabel('Capacity (Ah)'); plt.legend(); plt.show()
```
